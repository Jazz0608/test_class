```python
print(a + b)      # 加法:22
print(a - b)      # 減法:12
print(a * b)      # 乘法:85
print(a / b)      # 除法:3.4(浮點數)
print(a // b)     # 整數除法:3(17 除以 5 的商)
print(a % b)      # 取餘數:2(17 除以 5 的餘數)
print(a ** 2)     # 次方:289(17 的平方)
```

```python
a = 17
b = 5

print(a + b)      # 加法:22
print(a - b)      # 減法:12
print(a * b)      # 乘法:85
print(a / b)      # 除法:3.4(浮點數)
print(a // b)     # 整數除法:3(17 除以 5 的商)
print(a % b)      # 取餘數:2(17 除以 5 的餘數)
print(a ** 2)     # 次方:289(17 的平方)
```
