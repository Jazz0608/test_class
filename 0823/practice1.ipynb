```python
#如果 x 不是負數 , 則傳回值為 x ** (1/y)
#如果 x 是負數而且 y 為偶數 , 則傳回值為 "虛數"
#如果 x 是負數而且 y 為偶數 , 則傳回值為 -(-x) ** (1/y)

x = int(input('請輸入 x: '))
y = int(input('請輸入 y: '))
if x >= 0:
    print( x ** (1/y))
else:
    if y % 2 == 0:
        print("虛數")
    else:
        print(-(-x) ** (1/y))
```

```python
#以下是將學生分數換算成等級的程式, 對比這兩種寫法
score = int(input("請輸入分數:"))
if score >= 90:
    grade = "優"
else:
    if score >= 80:
        grade = "甲"
    else:
        if score >= 70:
            grade = "乙"
        else:
            if score >= 60:
                grade = "丙"
            else:
                grade = "丁"
print(grade)
```
