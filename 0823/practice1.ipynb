```python
#巢狀選擇(Nested it)
furry = True
small = False

if furry:
    if small:
        print("它是貓!")
    else:
        print("它是熊!")
else:
    if small:
        print("它是小蜥蜴!")
    else:
        print("它是人類或是沒毛的熊!")
                
```

```python
#如果 x 不是負數 , 則傳回值為 x ** (1/y)
#如果 x 是負數而且 y 為偶數 , 則傳回值為 "虛數"
#如果 x 是負數而且 y 為偶數 , 則傳回值為 -(-x) ** (1/y)

x = int(input('請輸入 x: '))
y = int(input('請輸入 y: '))
if x >= 0:
    print( x ** (1/y))
else:
    if y % 2 == 0:
        print("虛數")
    else:
        print(-(-x) ** (1/y))
```
